```python
import ee
import csv

ee.Initialize(project='cloud-removal-505010')  # <-- replace with your real project ID

DATE_START = '2023-11-01'
DATE_END   = '2024-02-28'
CLOUD_THRESH = 5

manifest = []

for idx, (name, coords) in enumerate(regions.items(), start=1):
    aoi = ee.Geometry.Rectangle(coords)

    # ---- Sentinel-2: filter + check candidates exist ----
    s2_col = (ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
              .filterBounds(aoi)
              .filterDate(DATE_START, DATE_END)
              .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', CLOUD_THRESH)))

    s2_count = s2_col.size().getInfo()
    if s2_count == 0:
        print(f"[{idx:02d}] {name}: SKIPPED - no clean S2 images in date range")
        continue

    s2_mosaic = s2_col.median()

    # ---- Sentinel-1: filter + check candidates exist ----
    s1_col = (ee.ImageCollection('COPERNICUS/S1_GRD')
              .filterBounds(aoi)
              .filterDate(DATE_START, DATE_END)
              .filter(ee.Filter.eq('instrumentMode', 'IW'))
              .filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV'))
              .filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VH'))
              .select(['VV', 'VH']))

    s1_count = s1_col.size().getInfo()
    if s1_count == 0:
        print(f"[{idx:02d}] {name}: SKIPPED - no S1 images in date range")
        continue

    s1_mosaic = s1_col.median()

    # ---- File names (paired by index) ----
    s2_name = f"image_{idx:02d}_s2"
    s1_name = f"image_{idx:02d}_s1"

    # ---- Export: S2 = RGB only (3 bands), S1 = VV+VH (2 bands) ----
    task_s2 = ee.batch.Export.image.toDrive(
        image=s2_mosaic.select(['B4', 'B3', 'B2']).clip(aoi),
        description=s2_name,
        folder='cloud_removal_dataset',
        scale=10, region=aoi, fileFormat='GeoTIFF', maxPixels=1e9
    )
    task_s1 = ee.batch.Export.image.toDrive(
        image=s1_mosaic.clip(aoi),
        description=s1_name,
        folder='cloud_removal_dataset',
        scale=10, region=aoi, fileFormat='GeoTIFF', maxPixels=1e9
    )
    task_s2.start()
    task_s1.start()

    manifest.append({
        'index': idx, 'region': name,
        's2_file': s2_name + '.tif', 's1_file': s1_name + '.tif',
        's2_candidates': s2_count, 's1_candidates': s1_count
    })

    print(f"[{idx:02d}] {name}: submitted as {s2_name} / {s1_name} "
          f"(S2 candidates={s2_count}, S1 candidates={s1_count})")

# ---- Save manifest locally so you always know region <-> filename mapping ----
with open('dataset_manifest.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['index','region','s2_file','s1_file','s2_candidates','s1_candidates'])
    writer.writeheader()
    writer.writerows(manifest)

print(f"\nTotal pairs submitted: {len(manifest)} / {len(regions)}")
print("Manifest saved to dataset_manifest.csv")
```
